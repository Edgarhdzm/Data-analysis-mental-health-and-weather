```python
import pandas as pd, numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib as mpl

from pathlib import Path
DATA = Path(r"C:\Users\clara\code\data_AV")
df = pd.read_csv(DATA / "usa_yearly.csv")

def fix(s):
    try:
        return s.encode("latin-1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return s

df["region_name"] = df["region_name"].map(fix)
df["daylight_h"] = df["daylight_duration"] / 3600

print(df.shape, df.isna().sum().sum())
print(df.describe().T)
```

```python
vars_ = ["temperature_2m_mean", "precipitation_hours", "rain_sum"]

# media nacional (simple, sin ponderar) + banda min-max entre regiones
fig, axs = plt.subplots(len(vars_), 1, figsize=(8, 8), sharex=True)
for ax, v in zip(axs, vars_):
    g = df.groupby("year")[v]
    ax.plot(g.mean(), color="k", lw=2)
    ax.fill_between(g.mean().index, g.min(), g.max(), alpha=0.2)
    ax.set_ylabel(v)
plt.tight_layout()

# small multiples: una línea por región
fig, axs = plt.subplots(4, 5, figsize=(16, 10), sharex=True)
for ax, (name, d) in zip(axs.flat, df.groupby("region_name")):
    ax.plot(d.year, d.temperature_2m_mean)
    ax.set_title(name, fontsize=8)
plt.tight_layout()
```

```python
v = "temperature_2m_mean"
piv = df.pivot(index="region_name", columns="year", values=v)
anom = piv.sub(piv.mean(axis=1), axis=0)          # anomalía respecto a la media 2001-2021
order = piv.mean(axis=1).sort_values().index      # ordenado de frío a cálido

fig, ax = plt.subplots(figsize=(10, 6))
im = ax.imshow(anom.loc[order], aspect="auto", cmap="RdBu_r",
               vmin=-1.5, vmax=1.5)
ax.set_yticks(range(len(order))); ax.set_yticklabels(order, fontsize=8)
ax.set_xticks(range(0, 21, 2)); ax.set_xticklabels(piv.columns[::2])
plt.colorbar(im, label="Anomalía T (°C)")
```

```python
import geopandas as gpd, numpy as np, pandas as pd
import matplotlib.pyplot as plt, matplotlib as mpl

GISCO = ("https://gisco-services.ec.europa.eu/distribution/v2/nuts/geojson/"
         "NUTS_RG_10M_2021_4326.geojson")
US_URL = "https://www2.census.gov/geo/tiger/GENZ2021/shp/cb_2021_us_state_20m.zip"

# descarga una vez y guarda en disco
nuts_path = DATA / "nuts.geojson"
if not nuts_path.exists():
    gpd.read_file(GISCO).to_file(nuts_path, driver="GeoJSON")
nuts = gpd.read_file(nuts_path)

us_path = DATA / "us_states.gpkg"
if not us_path.exists():
    gpd.read_file(US_URL).to_file(us_path, driver="GPKG")
us = gpd.read_file(us_path)

CRS = {"ES": 3035, "NL": 28992, "US": 5070}   # proyección apropiada de cada país

def load_geo(country):
    if country in ("ES", "NL"):
        g = nuts[(nuts.LEVL_CODE == 2) & (nuts.CNTR_CODE == country)]
        g = g[["NUTS_ID", "geometry"]].rename(columns={"NUTS_ID": "region_code"})
    else:
        # solo los 48 estados contiguos + DC (Alaska/Hawái/territorios fuera)
        out = ["AK", "HI", "PR", "VI", "GU", "MP", "AS"]
        g = us[~us.STUSPS.isin(out)][["STUSPS", "STATEFP", "NAME", "geometry"]]
        # AJUSTA esto a lo que tenga tu region_code (ver nota abajo)
        g = g.rename(columns={"STUSPS": "region_code"})
    return g.to_crs(CRS[country])
```

```python
def region_summary(df, vars_):
    rows = []
    for code, d in df.groupby("region_code"):
        r = {"region_code": code}
        for v in vars_:
            r[f"{v}_mean"] = d[v].mean()
            r[f"{v}_trend"] = np.polyfit(d.year, d[v], 1)[0] * 10   # por década
        rows.append(r)
    return pd.DataFrame(rows)

def add_anomalies(df, vars_):
    for v in vars_:
        df[f"{v}_anom"] = df[v] - df.groupby("region_code")[v].transform("mean")
    return df
```

```python
def plot_map(gdf, col, ax=None, title="", cmap="viridis", label="",
             center=None, vmin=None, vmax=None, country="ES", cbar=True):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 7))
    fig = ax.figure
    vmin = gdf[col].min() if vmin is None else vmin
    vmax = gdf[col].max() if vmax is None else vmax
    norm = (mpl.colors.TwoSlopeNorm(center, vmin, vmax) if center is not None
            else mpl.colors.Normalize(vmin, vmax))

    if country == "ES":
        main, can = gdf[gdf.region_code != "ES70"], gdf[gdf.region_code == "ES70"]
    else:
        main, can = gdf, None

    main.plot(column=col, ax=ax, cmap=cmap, norm=norm, edgecolor="white", linewidth=0.5)
    ax.set_axis_off(); ax.set_title(title, fontsize=10)

    if can is not None:
        axin = ax.inset_axes([0.0, 0.0, 0.3, 0.2])
        can.plot(column=col, ax=axin, cmap=cmap, norm=norm, edgecolor="white", linewidth=0.5)
        axin.set_axis_off()

    if cbar:
        sm = mpl.cm.ScalarMappable(norm=norm, cmap=cmap)
        fig.colorbar(sm, ax=ax, shrink=0.6, label=label)
    return ax
```

```python
vars_ = ["temperature_2m_mean", "precipitation_hours", "rain_sum", "daylight_h"]
df = add_anomalies(df, vars_)
gdf = load_geo("ES").merge(region_summary(df, vars_), on="region_code")

plot_map(gdf, "temperature_2m_mean_mean", title="T media 2001-2021",
         cmap="YlOrRd", label="°C")
plot_map(gdf, "temperature_2m_mean_trend", title="Tendencia de T",
         cmap="RdBu_r", label="°C/década", center=0)
plot_map(gdf, "rain_sum_mean", title="Lluvia anual media", cmap="Blues", label="mm")
```
