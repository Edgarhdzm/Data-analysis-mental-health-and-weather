```python

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import unicodedata
import re

# ---------------------------------------
# 1. CARGAR LOS DATOS DEL INE
# ---------------------------------------

df = pd.read_csv(
    "datos_ine_suicidio_ccaa_desde2001.csv",
    encoding="latin1",
    sep=None,
    engine="python"
)

# Seleccionar un año
year = 2012

df_year = df[df["Periodo"] == year].copy()

# Eliminar códigos numéricos al principio del nombre
df_year["CCAA"] = (
    df_year["Comunidades y Ciudades Autónomas"]
    .str.replace(r"^\d+\s+", "", regex=True)
)

# Eliminar registros que no corresponden a comunidades autónomas
df_year = df_year[df_year["CCAA"] != "Extranjero"]


# ---------------------------------------
# 2. NORMALIZAR LOS NOMBRES
# ---------------------------------------

# Adaptar los nombres del INE a los nombres geográficos
nombres_ine = {
    "Asturias, Principado de": "Principado de Asturias",
    "Balears, Illes": "Illes Balears",
    "Castilla - La Mancha": "Castilla-La Mancha",
    "Comunitat Valenciana": "Comunidad Valenciana",
    "Madrid, Comunidad de": "Comunidad de Madrid",
    "Murcia, Región de": "Región de Murcia",
    "Navarra, Comunidad Foral de": "Comunidad Foral de Navarra",
    "Rioja, La": "La Rioja"
}

df_year["CCAA"] = df_year["CCAA"].replace(nombres_ine)


# Función para comparar nombres ignorando acentos,
# mayúsculas y espacios
def normalizar(nombre):
    nombre = unicodedata.normalize("NFKD", str(nombre))
    nombre = "".join(
        c for c in nombre
        if not unicodedata.combining(c)
    )
    nombre = nombre.lower().strip()
    nombre = re.sub(r"[-_,]", " ", nombre)
    return " ".join(nombre.split())


df_year["join_name"] = df_year["CCAA"].apply(normalizar)


# ---------------------------------------
# 3. CARGAR LOS LÍMITES GEOGRÁFICOS
# ---------------------------------------

url = (
    "https://mapas.fomento.gob.es/arcgis/rest/services/"
    "SIU/ENTIDADES_TERRITORIALES_EGRN/MapServer/1/query"
    "?where=1%3D1&outFields=*&outSR=4326&f=geojson"
)

gdf = gpd.read_file(url)

# Comprobar las columnas del archivo geográfico
print(gdf.columns)


# ---------------------------------------
# 4. PREPARAR LA UNIÓN DE LOS DATOS
# ---------------------------------------

# El servicio geográfico utiliza NAMEUNIT
gdf["join_name"] = gdf["NAMEUNIT"].apply(normalizar)

# Unir geometrías y datos del INE
mapa = gdf.merge(
    df_year[["join_name", "Total"]],
    on="join_name",
    how="left"
)

# Comprobar si hay comunidades sin datos
print(mapa[["NAMEUNIT", "Total"]])


# ---------------------------------------
# 5. REPRESENTAR EL MAPA
# ---------------------------------------

fig, ax = plt.subplots(figsize=(12, 10))

mapa.plot(
    column="Total",
    cmap="Reds",
    linewidth=0.8,
    edgecolor="black",
    legend=True,
    legend_kwds={
        "label": "Número de suicidios",
        "orientation": "horizontal",
        "shrink": 0.7
    },
    missing_kwds={
        "color": "lightgrey",
        "label": "Sin datos"
    },
    ax=ax
)

ax.set_title(
    f"Suicidios por comunidad autónoma en España ({year})",
    fontsize=16,
    pad=20
)

ax.axis("off")

plt.tight_layout()
plt.show()
```
